```python
import matplotlib.pyplot as plt
import seaborn as sns

sns.countplot(x="Survived", data=df)
plt.title("Survival Count")
plt.show()
```

```python
sns.countplot(x="Survived", data=df)
plt.title("Survival Count")
plt.show()
```

```python
sns.countplot(x="Sex", data=df)
plt.title("Gender Distribution")
plt.show()
```

The graph shows that more passengers did not survive than survived.

```python
%pip install seaborn
```

```python
sns.histplot(df["Age"], bins=20)
plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Number of Passengers")
plt.show()
```

Most passengers are between 20 and 40 years old.

```python
%pip install matplotlib
```

```python
sns.countplot(x="Pclass", data=df)
plt.title("Passenger Class Distribution")
plt.show()
```

Most passengers traveled in third class.

```python
sns.countplot(x="Sex", hue="Survived", data=df)
plt.title("Survival by Gender")
plt.show()
```

Female passengers had a higher survival rate than male passengers.

```python
numeric_df = df.select_dtypes(include="number")

plt.figure(figsize=(8,6))
sns.heatmap(numeric_df.corr(), annot=True)
plt.title("Correlation Heatmap")
plt.show()
```

The heatmap shows the correlation between numerical variables such as Age, Fare, and Passenger Class.

```python
sns.boxplot(x=df["Fare"])
plt.title("Fare Box Plot")
plt.show()
```

The Fare column contains some outliers, indicating that a few passengers paid much higher fares than most others.
